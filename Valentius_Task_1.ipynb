```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.histplot(
    np.log1p(df["capacity_mw"].dropna()),
    kde=True,
    color="#2b5c8f",
    edgecolor="white",
    bins=30,
)

plt.title(
    "Distribution of Power Plant Capacity (Log Scale)", fontsize=14, pad=12
)
plt.xlabel("Log(Installed Capacity in MW + 1)", fontsize=11)
plt.ylabel("Number of Power Plants", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

3. Distribution of Power Plant Capacity (Log Scale)Description: This histogram with a KDE curve displays the distribution of installed capacity (capacity_mw) transformed onto a logarithmic scale ($\log(1 + \text{capacity\_mw})$). Transforming the axis expands the severe right skew of the raw data into a smooth, multi-modal distribution.                                       Inference: The log transformation reveals distinct peaks representing micro/distributed generation facilities (low log values) and utility-scale installations (higher log values). Log-transforming capacity_mw is effective for stabilizing variance prior to linear model training.

```python
avg_capacity = (
    df_eda.groupby('fuel1')['capacity_mw']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=avg_capacity.index,
    y=avg_capacity.values
)

plt.title('Average Power Plant Capacity by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Average Capacity (MW)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

4. Average Power Plant Capacity by Fuel TypeDescription: This bar chart compares the mean installed capacity (capacity_mw) across primary fuel categories (fuel1). Nuclear power facilities hold the highest average capacity per plant (>2,000 MW), followed by coal and natural gas.   Inference: Although solar and wind lead in overall plant counts, their average generation capacity per facility is among the lowest. Nuclear and fossil-fuel plants require large centralized footprints to achieve economies of scale and meet baseline energy demand.   

```python
import matplotlib.pyplot as plt

valid_trend = (
    df[
        (df["commissioning_year"].notna())
        & (df["commissioning_year"] >= 1900)
        & (df["commissioning_year"] <= 2018)
    ]
    .groupby("commissioning_year")
    .size()
    .reset_index(name="plant_count")
)

plt.figure(figsize=(14, 5))

plt.stem(
    valid_trend["commissioning_year"],
    valid_trend["plant_count"],
    linefmt="#2b5c8f",
    markerfmt="o",
    basefmt=" ",
)

plt.title(
    "Power Plant Commissioning Trend Over the Years (1900–2018)",
    fontsize=14,
    pad=12,
)
plt.xlabel("Commissioning Year", fontsize=11)
plt.ylabel("Number of Power Plants", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

5. Power Plant Commissioning Trend Over the Years (1900–2018)
Description: This stem plot tracks the number of power plants commissioned per year based on commissioning_year, filtered between 1900 and 2018. It illustrates steady baseline growth throughout the 20th century followed by rapid exponential expansion starting around 2000.
Inference: Global energy expansion accelerated rapidly in the 21st century,
driven primarily by the boom in renewable energy projects. Filtering out unrecorded (0 / NaN) commissioning entries eliminates artificial drops, presenting a clean historical timeline.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

num_cols = [
    "capacity_mw",
    "latitude",
    "longitude",
    "commissioning_year",
    "year_of_capacity_data",
    "generation_gwh_2016",
    "estimated_generation_gwh",
]

numeric_df = df[num_cols].copy()
numeric_df_filled = numeric_df.fillna(numeric_df.median())

corr = numeric_df_filled.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    cbar=True,
    vmin=-0.3,
    vmax=1.0,
)

plt.title("Correlation Heatmap of Numerical Features", fontsize=14, pad=12)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

7. Correlation Heatmap of Numerical FeaturesDescription: This correlation matrix displays Pearson correlation coefficients across all 7 numerical features in the dataset after handling missing values. It shows a strong positive correlation between capacity_mw and estimated_generation_gwh ($r = 0.88$) and a moderate correlation with generation_gwh_2016 ($r = 0.33$).   Inference: Nameplate plant capacity (capacity_mw) is the primary linear driver of expected total energy output. Geographical coordinates (latitude and longitude) exhibit strong mutual collinearity ($r = 0.99$) due to spatial clustering, but show negligible linear correlation with facility capacity.

```python
country_capacity = (
    df_eda.groupby('country_long')['capacity_mw']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=country_capacity.values,
    y=country_capacity.index
)

plt.title('Top 10 Countries by Total Installed Capacity')
plt.xlabel('Total Installed Capacity (MW)')
plt.ylabel('Country')

plt.tight_layout()
plt.show()
```

8. Top 10 Countries by Total Installed CapacityDescription: This horizontal bar chart ranks the top 10 countries by their cumulative installed generation capacity in megawatts (capacity_mw) using country_long. China and the United States lead globally by a substantial margin, followed by India and Japan.   Inference: Comparing this with Chart 1 demonstrates the difference between plant density and output. While the United States leads in total number of individual plant facilities, China exceeds the US in total cumulative installed capacity due to a higher concentration of mega-scale energy plants.

##Summary of Key Findings & Insights

1. **Power Infrastructure Density vs. Output Capacity:**
   - While the **United States** ranks highest in the total count of individual power plants, **China** leads in total installed generation capacity (MW). This indicates China's strategic focus on large-scale centralized mega-facilities compared to a more distributed network of smaller power units in the US.

2. **Dominance of Renewables vs. Baseline Capacity:**
   - Renewable fuel sources (**Hydro**, **Solar**, **Wind**) account for the largest number of individual plant locations worldwide. However, conventional generation sources—specifically **Nuclear** and **Coal**—maintain significantly higher average capacities per plant (>2,000 MW for Nuclear), showing that baseline grid power relies on concentrated utility-scale installations.

3. **Accelerated Growth in the 21st Century:**
   - Historical commissioning data indicates baseline growth throughout the 20th century, followed by an exponential surge in new plant installations starting around the year 2000, aligning with global acceleration in renewable energy adoption.

4. **Key Numerical Drivers of Energy Output:**
   - Installed capacity (`capacity_mw`) is the primary linear driver of expected total energy generation (`estimated_generation_gwh`, $r = 0.88$). Geographical coordinates (`latitude` and `longitude`) show high spatial clustering ($r = 0.99$) but no linear correlation with individual plant capacity.
